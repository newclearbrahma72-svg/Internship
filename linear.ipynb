```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)


x = np.linspace(1, 10, 100)
noise = np.random.normal(0, 3, 100)
y = 3*x + 4 + noise
plt.scatter(x, y,  color="purple", label="data points")

true_y = 3*x + 4
plt.plot(x, true_y, color="red", label="true line")
plt.legend()
plt.title("Linear Graph: y = 3x + 4 + noise")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python


import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

x = np.linspace(-10, 10, 80)
noise = np.random.normal(0, 5, 80)
y = x*x + 2*x + 1 + noise
plt.plot(x, y, 'o')

true_y = x*x + 2*x + 1
plt.plot(x, true_y, color="red", label="true curve")

plt.scatter(x, y)
plt.title("Non-Linear Graph: y = x²")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

r = 5
theta = np.linspace(0, 2*np.pi, 100)
noise_x = np.random.normal(0, 2, 100)
noise_y = np.random.normal(0, 2, 100)
x = r * np.cos(theta) + noise_x
y = r * np.sin(theta) + noise_y

plt.plot(x, y,'o')

true_x = r * np.cos(theta)
true_y = r * np.sin(theta)
plt.plot(true_x, true_y)

plt.scatter(x, y )
plt.title("Circular Graph")
plt.xlabel("X")
plt.ylabel("Y")
plt.axis("equal")
plt.grid(True)
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/newclearbrahma72-svg/Intership/main/IN_AS%20(2).csv"

df = pd.read_csv(url)

df.columns = df.columns.str.strip().str.lower()

df["date"] = pd.to_datetime(df["date"], errors="coerce")

assam = df[df["subregion1_name"] == "Assam"].copy()

assam = assam.dropna(subset=["date", "new_confirmed"])

assam = assam.sort_values("date")

assam["average"] = (
    assam["new_confirmed"].rolling(window=7).mean()
)

plt.figure(figsize=(12, 5))

plt.plot(
    assam["date"],
    assam["new_confirmed"],
    label="Daily Confirmed Cases",
    alpha=0.5
)

plt.plot(
    assam["date"],
    assam["average"],
    label="7-Day Moving Average",
    linewidth=2
)

plt.xlabel("Date")
plt.ylabel("New Cases")
plt.title("Daily COVID-19 Cases in Assam")

plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()


fig.add_trace(go.Scatter(
    x=assam["date"],
    y=assam["new_confirmed"],
    mode="lines",
    name="Daily Cases",
    line=dict(color="firebrick", width=3),

    hovertemplate=(
        "<b>%{x|%b %d %Y}</b><br>"
        "New Cases: %{y:,.0f}"
        "<extra></extra>"
    )
))


fig.update_traces(
    hoverlabel=dict(
        bgcolor="white",
        font_size=16,
        font_color="black"
    ),
    hoveron="points",
    selector=dict(type="scatter")
)

fig.update_layout(
    title="Daily COVID-19 Cases in Assam",
    xaxis_title="Date",
    yaxis_title="New Cases",
    template="plotly_white",
    hovermode="closest"
)

fig.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

plot_x = np.random.randint(0, 100, 50)
plot_y = np.random.randint(0, 100, 50)

plot_labels = np.where(plot_y < plot_x, 0, 1)
plt.figure(figsize=(9, 6))

plt.scatter(
    plot_x,
    plot_y,
    c=plot_labels,
    cmap="coolwarm",
    vmin=0,
    vmax=1,
    s=100,
    edgecolors="black"
)

plt.plot([0, 100], [0, 100], 'k--', label="Y = X")

plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.title("Binary Classification: X vs Y")

plt.colorbar(ticks=[0, 1], label="Class")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()


df = pd.DataFrame({
    "X": plot_x,
    "Y": plot_y,
    "Class": plot_labels
})

df.to_csv("XY_classification.csv", index=False)

from google.colab import files
files.download("XY_classification.csv")
```
